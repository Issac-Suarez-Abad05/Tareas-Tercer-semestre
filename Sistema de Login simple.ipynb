```python
# -*- coding: utf-8 -*-
"""
Sistema de Login simple - Pruebas de caja negra
Basado en la Actividad A: diseño de casos de prueba
Ejecutable en Google Colab
"""

import re

# ------------------------------------------------------------
# Función de validación (lógica del sistema bajo prueba)
# ------------------------------------------------------------
def validar_credenciales(usuario: str, password: str) -> tuple:
    """
    Simula la validación de login según las reglas:
    - Usuario: alfanumérico, longitud entre 4 y 15 caracteres.
    - Contraseña: mínimo 8 caracteres, al menos una mayúscula,
      un número y un carácter especial (cualquiera de !@#$%^&*).
    Retorna:
        (bool, str) -> (éxito, mensaje)
    """
    # Validar usuario
    if not re.match(r'^[A-Za-z0-9]{4,15}$', usuario):
        return (False, "Credenciales inválidas")  # mensaje genérico

    # Validar contraseña
    if len(password) < 8:
        return (False, "Credenciales inválidas")
    if not re.search(r'[A-Z]', password):
        return (False, "Credenciales inválidas")
    if not re.search(r'[0-9]', password):
        return (False, "Credenciales inválidas")
    if not re.search(r'[!@#$%^&*]', password):
        return (False, "Credenciales inválidas")

    # Si todo es correcto
    return (True, "Acceso concedido - Bienvenido")

# ------------------------------------------------------------
# Casos de prueba (datos de entrada y resultados esperados)
# ------------------------------------------------------------
test_cases = [
    {
        "id": "TC01 - Credenciales válidas",
        "usuario": "usuario123",
        "password": "Passw0rd!",
        "esperado_exito": True,
        "esperado_mensaje": "Acceso concedido - Bienvenido"
    },
    {
        "id": "TC02 - Usuario inválido (carácter !)",
        "usuario": "inválido!",
        "password": "Passw0rd!",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    },
    {
        "id": "TC03 - Contraseña corta",
        "usuario": "test_user",
        "password": "12345",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    },
    {
        "id": "TC04 - Contraseña sin mayúscula",
        "usuario": "test_user",
        "password": "password1!",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    },
    {
        "id": "TC05 - Contraseña sin número",
        "usuario": "test_user",
        "password": "Password!",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    },
    {
        "id": "TC06 - Contraseña sin especial",
        "usuario": "test_user",
        "password": "Passw0rd",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    },
    {
        "id": "TC07 - Usuario demasiado largo (16 caracteres)",
        "usuario": "usuario1234567890",  # 16 caracteres
        "password": "Passw0rd!",
        "esperado_exito": False,
        "esperado_mensaje": "Credenciales inválidas"
    }
]

# ------------------------------------------------------------
# Ejecución de pruebas y reporte
# ------------------------------------------------------------
def ejecutar_pruebas():
    print("=" * 70)
    print("EJECUCIÓN DE CASOS DE PRUEBA - SISTEMA DE LOGIN")
    print("=" * 70)

    total = len(test_cases)
    pasaron = 0
    fallaron = 0

    for caso in test_cases:
        exito, mensaje = validar_credenciales(caso["usuario"], caso["password"])
        resultado_ok = (exito == caso["esperado_exito"] and mensaje == caso["esperado_mensaje"])

        print(f"\nCaso: {caso['id']}")
        print(f"  Usuario: {caso['usuario']} | Contraseña: {caso['password']}")
        print(f"  Resultado obtenido -> éxito={exito}, mensaje='{mensaje}'")
        print(f"  Resultado esperado -> éxito={caso['esperado_exito']}, mensaje='{caso['esperado_mensaje']}'")

        if resultado_ok:
            print("  ✅ PRUEBA PASADA")
            pasaron += 1
        else:
            print("  ❌ PRUEBA FALLIDA")
            fallaron += 1

    print("\n" + "=" * 70)
    print(f"RESUMEN: {pasaron} pasadas, {fallaron} fallidas (total {total})")
    print("=" * 70)

if __name__ == "__main__":
    ejecutar_pruebas()
```
